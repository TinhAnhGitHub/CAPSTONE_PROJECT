```python
from videodeepsearch.evaluation.datasets import (
    DatasetBuilder,
)
import mlflow
import os

mlflow.set_tracking_uri(os.getenv("MLFLOW_TRACKING_URI", "http://100.113.186.28:5000"))
mlflow.set_experiment("vds-agent-validation")
```

```python
import json
file_dict = '/home/tinhanhnguyen/Desktop/HK8/Capstone/CAPSTONE_PROJECT/videodeepsearch/local/mlflow_eval_records.json'
with open(file_dict, 'r') as f:
    data_dict = json.load(f)
```

```python
dataset_builder = DatasetBuilder(
    name='HaystackDataset',
    description="This is just a subset of the dataset MAGNET Video Haystack, including topics cooking, publick speaking, and news",
    version="v0.0.4"
)
```

```python
from uuid import uuid4


for record in data_dict:
    record['inputs'].update(
        {
            'session_id': str(uuid4())
        }
    )
    dataset_builder.add_record_from_dict(record)
```

```python
dataset = dataset_builder.create_mlflow_dataset(
    experiment_id="vds-agent-validation-magnet",
    tags={
        'dataset': "Magnet: A Multi-agent Framework for Finding Audio-Visual Needles by Reasoning over Multi-Video Haystacks",
        'topics': "cooking-tutorials | public-speaking | news"
    }
)
```

```python
filter_datasets = mlflow.genai.datasets.search_datasets(filter_string="name = 'HaystackDataset'")[0]
# filter_datasets
# mlflow.genai.datasets.delete_dataset(dataset_id="d-d06c4ec6d9c34886b58f6c5be1e95548")
```
